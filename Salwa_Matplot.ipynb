```python
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns

bulan =['jan','feb','mar','apr','mei','juni','juli']
penjualan =[150, 200, 100, 250, 125, 299, 300]

plt.figure(figsize=(6,4))

sns.barplot (x=bulan, y=penjualan, palette='muted')

plt.plot(
    range(len(bulan)),
    penjualan,
    color='red',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='garis tren'
)

plt.title('Penjualan per bulan garis tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dan filter data Kota Medan
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)
mask = df.select_dtypes(include=['object']).apply(lambda col: col.str.contains('MEDAN', case=False, na=False)).any(axis=1)
df_medan = df[mask].copy()

num_cols = df_medan.select_dtypes(include=['number']).columns.tolist()
cat_cols = df_medan.select_dtypes(include=['object']).columns.tolist()

if len(num_cols) > 0:
    plt.figure(figsize=(10, 5))
    sns.set_theme(style="whitegrid")

    col_target = num_cols[0]
    time_col = next((c for c in cat_cols + num_cols if 'tahun' in c.lower() or 'year' in c.lower() or 'periode' in c.lower()), None)

    if time_col and time_col in df_medan.columns:
        df_sorted = df_medan.sort_values(by=time_col)
        plt.plot(df_sorted[time_col], df_sorted[col_target], marker='o', color='teal', linewidth=2.5)
        plt.title(f'Tren {col_target} - Kota Medan', fontsize=14, fontweight='bold')
        plt.xlabel(time_col, fontsize=12)
        plt.ylabel(col_target, fontsize=12)
        plt.xticks(rotation=45)
    else:
        sns.barplot(data=df_medan.head(10), x=df_medan.head(10).index, y=col_target, palette='Blues_r')
        plt.title(f'Top 10 Sampel Nilai {col_target} - Kota Medan', fontsize=14, fontweight='bold')
        plt.xlabel('Indeks Baris', fontsize=12)
        plt.ylabel(col_target, fontsize=12)

    plt.tight_layout()
    plt.show()
else:
    print("Kolom numerik tidak ditemukan.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)
mask = df.select_dtypes(include=['object']).apply(lambda col: col.str.contains('MEDAN', case=False, na=False)).any(axis=1)
df_medan = df[mask].copy()

num_cols = df_medan.select_dtypes(include=['number']).columns.tolist()
cat_cols = df_medan.select_dtypes(include=['object']).columns.tolist()

plt.figure(figsize=(9, 7))
sns.set_theme(style="whitegrid")

if len(num_cols) >= 3:
    corr_matrix = df_medan[num_cols[:5]].corr()
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', cbar=True, linewidths=0.5)
    plt.title('Matriks Korelasi Variabel Fiskal - Kota Medan', fontsize=14, fontweight='bold')
elif len(cat_cols) > 0:
    cat_target = cat_cols[0]
    val_counts = df_medan[cat_target].value_counts().head(5)
    plt.pie(val_counts, labels=val_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
    plt.title(f'Proporsi Kategori ({cat_target}) - Kota Medan', fontsize=14, fontweight='bold')
else:
    print("Data tidak mencukupi untuk membuat visualisasi ini.")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/refs/heads/main/dataset_fiskal_97_hipoFull.csv"
# Membaca data langsung menggunakan pandas tanpa pyodide
df = pd.read_csv(url)
mask = df.select_dtypes(include=['object']).apply(lambda col: col.str.contains('MEDAN', case=False, na=False)).any(axis=1)
df_medan = df[mask].copy()

num_cols = df_medan.select_dtypes(include=['number']).columns.tolist()
cat_cols = df_medan.select_dtypes(include=['object']).columns.tolist()

plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

if len(num_cols) >= 2:
    sns.scatterplot(data=df_medan, x=num_cols[0], y=num_cols[1], color='purple', alpha=0.7, s=80)
    plt.title(f'Korelasi: {num_cols[0]} vs {num_cols[1]} (Kota Medan)', fontsize=14, fontweight='bold')
    plt.xlabel(num_cols[0], fontsize=12)
    plt.ylabel(num_cols[1], fontsize=12)
elif len(cat_cols) > 0 and len(num_cols) >= 1:
    cat_target = cat_cols[0]
    top_cat = df_medan[cat_target].value_counts().head(5).index
    sub_df = df_medan[df_medan[cat_target].isin(top_cat)]
    sns.barplot(data=sub_df, x=cat_target, y=num_cols[0], palette='Set2')
    plt.title(f'Rata-rata {num_cols[0]} per {cat_target} (Kota Medan)', fontsize=14, fontweight='bold')
    plt.xticks(rotation=30)
    plt.xlabel(cat_target, fontsize=12)
    plt.ylabel(num_cols[0], fontsize=12)
else:
    print("Data tidak mencukupi untuk membuat visualisasi ini.")

plt.tight_layout()
plt.show()
```
